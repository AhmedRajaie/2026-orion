```python
import sys, os
while not os.path.isdir('src') and os.path.dirname(os.getcwd()) != os.getcwd():
    os.chdir('..')
sys.path.insert(0, 'src')

import numpy as np
import matplotlib.pyplot as plt
import json
from tradinglab.data_feed import DataFeed
from tradinglab.simulator import PortfolioSimulator
from tradinglab.backtester import run_backtest
from tradinglab.strategies.sma import sma_crossover_weights

feed = DataFeed.from_dir('data/egx', symbols=['COMI', 'HRHO', 'TMGH', 'SWDY', 'FWRY', 'ABUK'])
sim = PortfolioSimulator(feed, benchmark='egx30')
res = run_backtest(sim, lambda o: sma_crossover_weights(o, 9, 20), lookback=30)

print('what /backtest will serve:')
preview = {
    'dates': [d.strftime('%Y-%m-%d') for d in res['dates'][-60:]],
    'portfolio': list((res['portfolio'][-60:] * 1000).round(2)),
    'benchmark': list((res['benchmark'][-60:] * 1000).round(2)),
}
print(json.dumps(preview)[:250], '...')
```
